```python
from themealdb_client import TheMealDBClient
```

```python
def test_search_recipes_by_ingredients():
    client = TheMealDBClient()
    ingredients = ["chicken","Basmati Rice"]
    recipes = client.search_recipes_by_ingredients(ingredients)
    print(f"Recipes with ingredient '{ingredients}':")
    for recipe in recipes:
        print(f"- {recipe['strMeal']}")

test_search_recipes_by_ingredients()
```

```python
def test_args(ingredient, *args ):
    print(f"Ingredient: {ingredient}")
    print(f"Additional args: {args}")
test_args(["tomato", "extra1"] )
    
    
```

```python
def test_get_all_ingredients():
    client = TheMealDBClient()
    ingredients = client.get_all_ingredients()
    print("All ingredients:")
    for ingredient in ingredients:
        print(f"- {ingredient['strIngredient']}")
test_get_all_ingredients()
```
