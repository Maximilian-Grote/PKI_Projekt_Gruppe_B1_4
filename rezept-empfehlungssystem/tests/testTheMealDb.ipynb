```python
import sys
import os

# Pfad zum übergeordneten Verzeichnis hinzufügen (damit themealdb_client gefunden wird)
sys.path.insert(0, os.path.abspath('..'))

from themealdb_client import TheMealDBClient
import json
import traceback

# Test für get_all_ingredients() Funktion

# Client initialisieren
client = TheMealDBClient()

# Funktion ausführen
print("=== Test: get_all_ingredients() ===\n")
try:
    ingredients = client.get_all_ingredients()
    print(f"Rückgabewert (ingredients): {(ingredients)[:10]}\n")
    str_ingredients = [ingredient['strIngredient'] for ingredient in ingredients if 'strIngredient' in ingredient]
    
    # Debug-Informationen
    print(f"✓ Anzahl der Zutaten: {len(ingredients)}")
    print(f"✓ Typ des Rückgabewerts: {type(ingredients)}\n")
    
    # Erste 5 Zutaten anzeigen
    print("--- Erste 5 Zutaten ---")
    for i, ingredient in enumerate(str_ingredients[:5], 1):
        print(f"{i}. {ingredient}")
        print(f"   Typ: {type(ingredient)}")
        if isinstance(ingredient, dict):
            print(f"   Keys: {ingredient.keys()}")
        print()
    
    # Prüfen ob Bild-URLs korrekt generiert werden können
    print("--- Beispiel Bild-URLs ---")
    for ingredient in ingredients[:3]:
        if isinstance(ingredient, dict) and 'strIngredient' in ingredient:
            name = ingredient['strIngredient']
            url = f"https://www.themealdb.com/images/ingredients/{name}.png"
            print(f"{name}: {url}")
    
    print("\n✓ Test erfolgreich abgeschlossen!")
    
except Exception as e:
    print(f"✗ Fehler beim Testen: {type(e).__name__}")
    print(f"  Nachricht: {str(e)}")
    print("\n--- Vollständiger Traceback ---")
    traceback.print_exc()

# Debugging-Ausgabe für Pfade
print("\n--- Debug: Pfadinformationen ---")
print(f"Aktuelles Arbeitsverzeichnis: {os.getcwd()}")
print(f"Python-Pfade (sys.path):")
for path in sys.path[:5]:  # Nur erste 5 anzeigen
    print(f"  - {path}")

# Versuche das Modul im übergeordneten Verzeichnis zu finden
parent_dir = os.path.abspath('..')
print(f"\nÜbergeordnetes Verzeichnis: {parent_dir}")
print(f"Inhalt des übergeordneten Verzeichnisses:")
if os.path.exists(parent_dir):
    for item in os.listdir(parent_dir):
        print(f"  - {item}")
```
